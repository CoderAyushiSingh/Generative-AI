##Question 1: What is a Vector Database (VectorDB) and how is it different from traditional databases?

- A Vector Database (VectorDB) is a database designed to store and search data in the form of vectors (numerical embeddings) instead of simple rows and columns. These vectors are usually generated from text, images, audio, etc., using AI models, and they capture the meaning or context of the data.

- It's different because:-

    1. Data Storage

          - Traditional DB: Stores structured data (tables, rows, columns like SQL).
          - VectorDB: Stores high-dimensional vectors
    2. Search Method

         - Traditional DB: Uses exact match queries      

        -  VectorDB: Uses similarity search     
    3. Use Case

        - Traditional DB: Best for structured data (bank records, user info).
        - VectorDB: Best for AI-based tasks    
    4. Query Type

        - Traditional DB: “Give me exact data.”
        - VectorDB: “Give me data that is similar in meaning.”    

##Question 2: Explain the various types of VectorDBs available and describe their suitability for different use cases.        

1. Standalone Vector Databases (Purpose-built)

   - These are built only for vector search.

    - Examples: Pinecone, Weaviate, Milvus

    - Features:

       - Highly optimized for similarity search (ANN algorithms like HNSW, IVF)
        - Scalable and fast
        - Built-in indexing and filtering
2. Vector Support in Traditional Databases

   - These are normal databases upgraded with vector capabilities.

    - Examples: PostgreSQL, MongoDB

    - Features:

       - Combines structured + vector data
       - Easier integration (since companies already use them)
       - Not as fast as dedicated VectorDBs        

3. In-Memory Vector Databases

   - These store vectors in RAM for ultra-fast access.

   - Examples: Redis

   - Features:

      - Extremely fast (low latency)
      - Limited by RAM size
      - Temporary or cache-like storage

4. Embedded / Local Vector Databases

    - Run locally inside your app — no server needed.

    - Examples: FAISS, Chroma

    - Features:

       - Lightweight
       - No cloud dependency
       - Easy to experiment
5. Hybrid Vector Databases

    - Combine keyword search + vector search.

     - Examples: Elasticsearch, OpenSearch

     - Features:

         - Combines BM25 (keyword) + semantic search
         - Better search accuracy
        - Supports filters + ranking

##Question 6: Install and set up Chroma DB, and insert sample vector data for semantic search.






```python
!pip install chromadb
import chromadb
from chromadb.utils import embedding_functions

# Create client (local DB)
client = chromadb.Client()

# Use default embedding (or you can use HuggingFace/OpenAI later)
embedding_function = embedding_functions.DefaultEmbeddingFunction()

# Create collection
collection = client.create_collection(
    name="my_collection",
    embedding_function=embedding_function
)

# Sample data
documents = [
    "I love machine learning",
    "Artificial intelligence is the future",
    "I enjoy playing football",
    "Data science is amazing"
]

# Insert data
collection.add(
    documents=documents,
    ids=["1", "2", "3", "4"]
)

print("Data inserted successfully!")

# Perform semantic search
query = "AI and ML"
results = collection.query(
    query_texts=[query],
    n_results=2
)

print("\nSearch Results:")
print(results)
```

##Question 7: Demonstrate how to download and fine-tune a Hugging Face model for a text generation task.


```python
# Install (run this in Colab)
!pip install transformers datasets accelerate

# Imports
from transformers import AutoTokenizer, AutoModelForCausalLM, Trainer, TrainingArguments
from datasets import Dataset

# 1. Load model + tokenizer (Hugging Face)
model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# IMPORTANT FIX for GPT-2
tokenizer.pad_token = tokenizer.eos_token

# 2. Sample dataset
data = {
    "text": [
        "Machine learning is transforming the world.",
        "Artificial intelligence will shape the future.",
        "Data science helps in making better decisions."
    ]
}

dataset = Dataset.from_dict(data)

# 3. Tokenization + LABEL FIX (THIS WAS YOUR ERROR)
def tokenize_function(example):
    tokens = tokenizer(
        example["text"],
        padding="max_length",
        truncation=True,
        max_length=50
    )

    # FIX: add labels
    tokens["labels"] = tokens["input_ids"].copy()

    return tokens

tokenized_dataset = dataset.map(tokenize_function, batched=True)

# IMPORTANT: set format for PyTorch
tokenized_dataset.set_format(
    type="torch",
    columns=["input_ids", "attention_mask", "labels"]
)

# 4. Training arguments
training_args = TrainingArguments(
    output_dir="./results",
    per_device_train_batch_size=2,
    num_train_epochs=3,
    logging_steps=5,
    save_steps=10,
    save_total_limit=2,
    report_to="none"  # avoids warnings in Colab
)

# 5. Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset
)

# 6. Train
trainer.train()

# 7. Save model
model.save_pretrained("./fine_tuned_model")
tokenizer.save_pretrained("./fine_tuned_model")

# 8. Test the model
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="./fine_tuned_model",
    tokenizer="./fine_tuned_model"
)

result = generator("AI is", max_length=30)
print(result)
```

##Question 8: Create a custom LLM using Ollama and Llama2, and run it locally for basic text prompts.


```python
# Make sure Ollama is installed and running, and you have run:
# ollama pull llama2
# ollama create my-llm -f Modelfile   (optional custom model)

import requests

# Function to query Ollama model
def run_llm(prompt, model="llama2"):
    url = "http://localhost:11434/api/generate"

    data = {
        "model": model,
        "prompt": prompt,
        "stream": False
    }

    response = requests.post(url, json=data)
    return response.json()["response"]

# Test with base LLaMA2 model
print("Base Model Output:\n")
print(run_llm("Explain artificial intelligence in simple terms", model="llama2"))

# Test with custom model (if created)
print("\nCustom Model Output:\n")
print(run_llm("What is data science?", model="my-llm"))
```

##Question 9: Implement a basic RAG (Retrieval-Augmented Generation) system using Ollama with Llama3.



```python
# Make sure you have:
# 1. Ollama running locally
# 2. Run in terminal:
#    ollama pull llama3
#    ollama pull nomic-embed-text
#
# Install dependencies:
# !pip install chromadb requests

import chromadb
import requests

# -----------------------------
# Step 1: Create Chroma DB
# -----------------------------
client = chromadb.Client()
collection = client.create_collection(name="rag_collection")

# -----------------------------
# Step 2: Sample Documents
# -----------------------------
documents = [
    "Machine learning is a subset of artificial intelligence.",
    "Deep learning uses neural networks with many layers.",
    "Natural language processing helps machines understand human language.",
    "Data science involves statistics, data analysis, and machine learning."
]

ids = [str(i) for i in range(len(documents))]

# -----------------------------
# Step 3: Generate embeddings using Ollama
# -----------------------------
def get_embedding(text):
    response = requests.post(
        "http://localhost:11434/api/embeddings",
        json={
            "model": "nomic-embed-text",
            "prompt": text
        }
    )
    return response.json()["embedding"]

embeddings = [get_embedding(doc) for doc in documents]

# Store in ChromaDB
collection.add(
    documents=documents,
    embeddings=embeddings,
    ids=ids
)

# -----------------------------
# Step 4: RAG Query Function
# -----------------------------
def rag_query(query):
    query_embedding = get_embedding(query)

    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=2
    )

    retrieved_docs = results["documents"][0]

    # Combine retrieved context
    context = "\n".join(retrieved_docs)

    prompt = f"""
    Use the following context to answer the question:

    {context}

    Question: {query}
    Answer:
    """

    # Generate response using LLaMA3
    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": "llama3",
            "prompt": prompt,
            "stream": False
        }
    )

    return response.json()["response"]

# -----------------------------
# Step 5: Test RAG
# -----------------------------
query = "What is machine learning?"
answer = rag_query(query)

print("Query:", query)
print("Answer:", answer)
```
